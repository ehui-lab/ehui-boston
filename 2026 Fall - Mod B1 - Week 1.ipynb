```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import interact, FloatSlider, IntSlider
from sklearn.linear_model import LinearRegression


def plot_advertising_sales_regression(
    n_stores=50,
    sales_lift_per_1000_dollars=8.0,
    noise=10.0,
    baseline_sales=40.0,
    random_seed=42
):
    rng = np.random.default_rng(random_seed)

    ad_spend_thousands = rng.uniform(0, 20, size=n_stores)

    weekly_sales_thousands = (
        baseline_sales
        + sales_lift_per_1000_dollars * ad_spend_thousands
        + rng.normal(0, noise, size=n_stores)
    )

    X = ad_spend_thousands.reshape(-1, 1)
    y = weekly_sales_thousands

    model = LinearRegression()
    model.fit(X, y)

    ad_spend_line = np.linspace(0, 20, 200)
    true_sales_line = baseline_sales + sales_lift_per_1000_dollars * ad_spend_line
    modeled_sales_line = model.intercept_ + model.coef_[0] * ad_spend_line

    plt.figure(figsize=(9, 6))

    plt.scatter(
        ad_spend_thousands,
        weekly_sales_thousands,
        alpha=0.75,
        label="Stores / weekly observations"
    )

    plt.plot(
        ad_spend_line,
        true_sales_line,
        color="green",
        linewidth=3,
        label=(
            "True business relationship: "
            f"Sales = {baseline_sales:.1f} + "
            f"{sales_lift_per_1000_dollars:.1f} × Ad Spend"
        )
    )

    plt.plot(
        ad_spend_line,
        modeled_sales_line,
        color="red",
        linestyle="--",
        linewidth=3,
        label=(
            "Linear regression estimate: "
            f"Sales = {model.intercept_:.1f} + "
            f"{model.coef_[0]:.1f} × Ad Spend"
        )
    )

    plt.title("Advertising Spend vs. Weekly Sales")
    plt.xlabel("Weekly Advertising Spend ($1,000s)")
    plt.ylabel("Weekly Sales Revenue ($1,000s)")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()


interact(
    plot_advertising_sales_regression,
    n_stores=IntSlider(
        value=50,
        min=5,
        max=300,
        step=5,
        description="Stores"
    ),
    sales_lift_per_1000_dollars=FloatSlider(
        value=8.0,
        min=-5.0,
        max=20.0,
        step=0.5,
        description="Sales lift"
    ),
    noise=FloatSlider(
        value=10.0,
        min=0.0,
        max=100.0,
        step=1.0,
        description="Noise"
    ),
    baseline_sales=FloatSlider(
        value=40.0,
        min=0.0,
        max=100.0,
        step=5.0,
        description="Baseline"
    ),
    random_seed=IntSlider(
        value=42,
        min=0,
        max=100,
        step=1,
        description="Seed"
    )
);

# Supervised
```

# Supervised Learning (above)

```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import interact, IntSlider, FloatSlider
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler


def plot_customer_kmeans_segmentation(
    n_customers=250,
    n_clusters=3,
    spending_noise=0.8,
    visits_noise=0.8,
    random_seed=42
):
    rng = np.random.default_rng(random_seed)

    segment_centers = np.array([
        [25, 2],
        [75, 6],
        [140, 10],
        [220, 4],
        [55, 12]
    ])

    true_segment_ids = rng.integers(
        low=0,
        high=len(segment_centers),
        size=n_customers
    )

    monthly_spend = (
        segment_centers[true_segment_ids, 0]
        + rng.normal(0, spending_noise * 15, size=n_customers)
    )

    monthly_visits = (
        segment_centers[true_segment_ids, 1]
        + rng.normal(0, visits_noise * 1.5, size=n_customers)
    )

    monthly_spend = np.clip(monthly_spend, 5, 300)
    monthly_visits = np.clip(monthly_visits, 1, 20)

    X = np.column_stack([monthly_spend, monthly_visits])

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    kmeans = KMeans(
        n_clusters=n_clusters,
        random_state=random_seed,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_scaled)

    cluster_centers_original_units = scaler.inverse_transform(
        kmeans.cluster_centers_
    )

    x_min = monthly_spend.min() - 20
    x_max = monthly_spend.max() + 20
    y_min = monthly_visits.min() - 2
    y_max = monthly_visits.max() + 2

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 500),
        np.linspace(y_min, y_max, 500)
    )

    grid_points_original_units = np.column_stack([
        xx.ravel(),
        yy.ravel()
    ])

    grid_points_scaled = scaler.transform(grid_points_original_units)

    grid_cluster_labels = kmeans.predict(grid_points_scaled)
    grid_cluster_labels = grid_cluster_labels.reshape(xx.shape)

    colors = plt.cm.tab10(np.arange(n_clusters))

    plt.figure(figsize=(10, 7))

    plt.contourf(
        xx,
        yy,
        grid_cluster_labels,
        levels=np.arange(n_clusters + 1) - 0.5,
        colors=colors,
        alpha=0.15
    )

    plt.contour(
        xx,
        yy,
        grid_cluster_labels,
        levels=np.arange(n_clusters + 1) - 0.5,
        colors="black",
        linestyles="dashed",
        linewidths=1.5,
        alpha=0.8
    )

    for cluster_id in range(n_clusters):
        mask = cluster_labels == cluster_id

        plt.scatter(
            monthly_spend[mask],
            monthly_visits[mask],
            color=colors[cluster_id],
            alpha=0.75,
            s=55,
            edgecolor="black",
            linewidth=0.4,
            label=f"Cluster {cluster_id}"
        )

    plt.scatter(
        cluster_centers_original_units[:, 0],
        cluster_centers_original_units[:, 1],
        c="red",
        marker="X",
        s=275,
        edgecolor="black",
        linewidth=1.5,
        label="Cluster centers"
    )

    plt.title(
        "Customer Segmentation with K-Means\n"
        "Voronoi Regions Around Each Cluster Center"
    )

    plt.xlabel("Monthly Customer Spend ($)")
    plt.ylabel("Monthly Store Visits")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    print("The dashed boundaries separate the Voronoi regions.")
    print("Each region contains the customers closest to one K-Means center.")
    print()
    print("Estimated customer segment centers:")

    for i, center in enumerate(cluster_centers_original_units):
        print(
            f"Cluster {i}: "
            f"${center[0]:.2f} monthly spend, "
            f"{center[1]:.1f} visits/month"
        )


interact(
    plot_customer_kmeans_segmentation,
    n_customers=IntSlider(
        value=250,
        min=50,
        max=1000,
        step=50,
        description="Customers"
    ),
    n_clusters=IntSlider(
        value=3,
        min=2,
        max=8,
        step=1,
        description="K clusters"
    ),
    spending_noise=FloatSlider(
        value=0.8,
        min=0.1,
        max=3.0,
        step=0.1,
        description="Spend noise"
    ),
    visits_noise=FloatSlider(
        value=0.8,
        min=0.1,
        max=3.0,
        step=0.1,
        description="Visit noise"
    ),
    random_seed=IntSlider(
        value=42,
        min=0,
        max=100,
        step=1,
        description="Seed"
    )
);
```
