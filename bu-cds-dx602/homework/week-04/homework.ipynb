**Concept:** *Functions with lists and indexing.*

**Task:**

Write a function middlest_element(values) that returns the middle element of a list.

Rules:

* If the list is empty, return `None`.

* If the list length is odd, return the element at the exact middle index.

* If the list length is even, return the element just left of center (example: for length 6, use index 2).

* Then run the provided test cases and print results exactly as shown.

**Sample Output:**

```
middlest_element([1,2,3,4,5]) =  3
middlest_element([1,2,3,4,5,6]) =  3
middlest_element([]) =  None
```


````python
def middlest_element(values):
    """Return the middle element, or the left-middle element for even lengths."""
    # An empty list has no middle element.
    if len(values) == 0:
        return None
    # Integer division selects the exact middle for odd lengths and left-middle for even lengths.
    return values[(len(values) - 1) // 2]
````

**Review Problem 2**

**Concept:** *List comprehensions.*

**Task:**

Write a function `square_evens(values)` that returns a new list containing the square of every even number in `values`, in the same order they appear.

Rules:

* Must use a list comprehension.

* Then write your own test cases that clearly prove correctness (include at least: mixed evens/odds, all odds, empty list, negative evens).

**Sample Output:**

```
square_evens([3, 4, 2, 6, 9, 8])
```
should return

`[16, 4, 36, 64]`

````python
def square_evens(values):
    """Return the squares of even values, preserving their original order."""
    # Keep even values and square each one in the list comprehension.
    return [value ** 2 for value in values if value % 2 == 0]
````

````python
def print_even_odd(values):
    """Print whether each value in the list is even or odd."""
    # Visit each value directly rather than using list indices.
    for value in values:
        # Use the remainder to decide whether the value is even or odd.
        if value % 2 == 0:
            print(f"{value} is even")
        else:
            print(f"{value} is odd")
````

````python
def print_names(names):
    """Print each name on its own line."""
    # Use each index to access and print its name.
    for i in range(len(names)):
        print(names[i])


def print_adjacent_names(names):
    """Print each name alongside the next name in the list."""
    # Stop before the final index so the next-name lookup stays in bounds.
    for i in range(len(names) - 1):
        print(f"{names[i]} , {names[i + 1]}")
````

**Problem 6**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `mean(values)` that returns the arithmetic mean of a list of numbers.

Rules:

* Must use a loop + accumulator (no `sum()`).

* No imports.

* Then run the given test case and print exactly in the same format.


**Sample Output:**

```
values: [2, 1, 6, 3, 7]
mean: 3.8
```


````python
def mean(values):
    """Return the arithmetic mean of a non-empty list of numbers."""
    # Return None when there are no values to average.
    if len(values) == 0:
        return None

    total = 0
    # Accumulate each value without using sum().
    for value in values:
        total += value
    # Divide the accumulated total by the number of values.
    return total / len(values)
````

````python
values = [2, 1, 6, 3, 7]
print(f"values: {values}")
print(f"mean: {mean(values)}")
````

**Problem 7**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `variance(values)` that returns the population variance of a list of numbers using this formula:

$$
\text{variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

Rules:

* Must use loops + accumulator (no `sum()`).

* No imports.

* Must call your `mean(values)` function to get the mean.

* Then run the provided test case and print in the same format.

**Sample Output:**

```
values: [6, 3, 7, 9, 5]
mean: 6.0
variance: 4.0
```


````python
def variance(values):
    """Return the population variance of a non-empty list of numbers."""
    # Use the mean function to find the list's average.
    average = mean(values)
    # Return None when there is no average for an empty list.
    if average is None:
        return None

    squared_deviations = 0
    # Accumulate each squared difference from the mean.
    for value in values:
        squared_deviations += (value - average) ** 2
    # Divide by the population size to compute population variance.
    return squared_deviations / len(values)
````

````python
values = [6, 3, 7, 9, 5]
print(f"values: {values}")
average = mean(values)
print(f"mean: {average}")
print(f"variance: {variance(values)}")
````

**Problem 8**

**Concept:** *Accumulator pattern with numeric result, list iteration, index-based loop.*

**Task:**

Write a function `covariance(xs, ys)` that computes the population covariance between two series:

$$
\text{covariance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})
$$

Rules:

* If `xs` and `ys` are not the same length, print:
    
    `xs and ys must be the same length`
    
    and return `0`.

* Must use an index-based loop.

* Must use your `mean()` function.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```
x values: [6, 3, 7, 9, 5]
y values: [4, 1, 7, 5, 0]
covariance: 3.8
```


````python
def covariance(xs, ys):
    """Return the population covariance of two equal-length lists."""
    # Stop if the paired lists do not have the same number of values.
    if len(xs) != len(ys):
        # Explain why the function cannot calculate covariance.
        print("xs and ys must be the same length")
        return 0
    # An empty pair has no population covariance to average.
    if len(xs) == 0:
        return 0

    # Calculate the average of each input list.
    x_mean = mean(xs)
    y_mean = mean(ys)
    total = 0
    # Use matching indices to accumulate products of deviations.
    for i in range(len(xs)):
        total += (xs[i] - x_mean) * (ys[i] - y_mean)
    # Divide by the population size to compute population covariance.
    return total / len(xs)
````

Test cases for Problem 8

````python
xs = [6, 3, 7, 9, 5]
ys = [4, 1, 7, 5, 0]
print(f"x values: {xs}")
print(f"y values: {ys}")
print(f"covariance: {covariance(xs, ys)}")

# Check the required behavior for lists with different lengths.
print(covariance([1, 2], [1]))
````

Write a function `polynomial_value(x, coefficients)` that evaluates a polynomial where each element in `coefficients` is the coefficient for

Example meaning:

Rules:

````python
def polynomial_value(x, coefficients):
    """Evaluate a polynomial from its coefficients in ascending powers."""
    result = 0
    # Each coefficient multiplies x raised to its index.
    for i in range(len(coefficients)):
        result += coefficients[i] * (x ** i)
    return result
````

````python
def string_to_ascii(s):
    """Return a list of ASCII codes for the characters in s."""
    codes = []
    # Convert each character and append its code to the result list.
    for character in s:
        codes.append(ord(character))
    return codes
````

**Problem 12**

**Concept:** *Accumulator pattern to process a string, produce a string result.*

**Task:**

Write a function `cipher(s, n)` that applies a Caesar cipher shift to a plaintext string.

Rules:

* Assume lowercase letters only for shifting (`a` to `z`).

* Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

* Must wrap around the alphabet if shifting past `z` or before `a`.

* Must use an accumulator string.

* Then run the exact test cases shown and print in the same format.

**Sample Output:**

```
plaintext = abc, ciphered = def
plaintext = abc xyz, ciphered = def abc
```


````python
def cipher(s, n):
    """Return s shifted by n places using a lowercase Caesar cipher."""
    result = ""
    # Process each character, leaving non-lowercase letters unchanged.
    for character in s:
        # Only shift lowercase letters a through z.
        if "a" <= character <= "z":
            # Convert to a 0-25 position, shift by n, and wrap around with % 26.
            shifted_index = (ord(character) - ord("a") + n) % 26
            # Convert the shifted position back into a letter and add it to the result.
            result += chr(ord("a") + shifted_index)
        else:
            # Spaces and punctuation are added unchanged.
            result += character
    return result
````

Test case for problem 12.

````python
plaintext = "abc"
result = cipher(plaintext, 3)
print(f"plaintext = {plaintext}, ciphered = {result}")

plaintext = "abc xyz"
result = cipher(plaintext, 3)
print(f"plaintext = {plaintext}, ciphered = {result}")
````

**Problem 13**

**Concept:** *More practice with loops.*

**Task:**

Write a function `print_all_decipherings(s)` that prints all 26 possible Caesar shifts of string `s`.

Rules:

* Use a definite loop for shifts `0` through `25`.

* For each shift `i`, compute `ciphered = cipher(s, i)` and print:

    `{i}, {ciphered}`

* Then run these two test cases:

    `s = "def"`

    `s = "oqfcgg hvs ibwjsfgs"`

**Sample Output:**

For `"def"` it must print 26 lines starting with `0, def` and ending with `25, cde`

For `"oqfcgg hvs ibwjsfgs"` it must print 26 lines. One of theses is the name of a Beatles song.



````python
def print_all_decipherings(s):
    """Print all 26 Caesar shifts of s."""
    # Try every possible shift from zero through twenty-five.
    for i in range(26):
        # Call cipher to shift the string by i places.
        deciphered = cipher(s, i)
        print(f"{i}, {deciphered}")
````
