**Review Problem 2**

**Concept:** *reading CSV data, matplotlib, line graph*

**Task:**

Write a function `read_and_plot_column(filename, column_name)` that:
* Opens a CSV file (do **not** use pandas)
* Extracts the values from the specified column
* Casts all values to `float`
* Plots the values using `plt.plot()`
* Adds a title
* Displays the plot

````python
def read_and_plot_column(filename, column_name):
    """Open a CSV file without pandas, plot one numeric column, and show the chart."""
    def split_csv_line(line):
        """Split one CSV line on commas that are not inside quotes."""
        fields = []
        current = []
        in_quotes = False
        for char in line:
            # A quote toggles whether the following commas are separators.
            if char == '"':
                in_quotes = not in_quotes
            elif char == "," and not in_quotes:
                fields.append("".join(current))
                current = []
            else:
                current.append(char)
        fields.append("".join(current))
        return fields

    with open(filename, "r") as file:
        # The first row holds the column names.
        headers = split_csv_line(file.readline().strip())
        column_index = headers.index(column_name)
        values = []
        # Read each remaining row and keep the requested column.
        for line in file:
            if line.strip() == "":
                continue
            text = split_csv_line(line.strip())[column_index]
            # Blank cells become missing values; everything else is a float.
            if text.strip() == "":
                values.append(np.nan)
            else:
                values.append(float(text))
    plt.plot(values)
    plt.title(column_name)
    plt.xlabel("Row")
    plt.ylabel(column_name)
    plt.show()


````

**Task:**

Using `df_weather`:
* Write a helper function to extract time (in hours) from a datetime string and convert it to decimal hours (float)
* Convert time using:

    `hours + (minutes / 60) + (seconds / 3600)`
* Create a new DataFrame with the same index as `df_weather`
* Add the following columns:
  * `sunrise` (converted to float)
  * `sunset` (converted to float)
  * `tempmin` (copied directly from `df_weather`)
* Plot all columns together using `.plot()`


**Sample Output:**

This is similar to how your new dataframe should look before plotting:

```
   start_time  end_time  price
P1       8.25     17.50     20
P2       8.50     17.75     35
P3       9.00     18.00     15
P4       9.25     18.25     60
P5       9.50     18.50     50
```

````python
def decimal_hours(datetime_string):
    """Convert a datetime string to decimal hours."""
    text = str(datetime_string).strip()
    # Keep the clock time when a date is included.
    if "T" in text:
        text = text.split("T", 1)[1]
    elif " " in text:
        text = text.split(" ", 1)[1]
    hours_text, minutes_text, seconds_text = text.split(":")
    # hours + (minutes / 60) + (seconds / 3600)
    return float(hours_text) + (float(minutes_text) / 60) + (float(seconds_text) / 3600)


# New frame shares the weather dates, then stores the converted columns.
df_sun = pd.DataFrame(index=df_weather.index)
df_sun["sunrise"] = df_weather["sunrise"].apply(decimal_hours)
df_sun["sunset"] = df_weather["sunset"].apply(decimal_hours)
df_sun["tempmin"] = df_weather["tempmin"]
print(df_sun.head())
df_sun.plot(title="Sunrise, Sunset, and Minimum Temperature")
plt.show()



````

**Problem 15**

**Concept:** *Filtering `DataFrame` with boolean conditions*


**Task:**


Write a function `filter_rows_matching_pattern(df, column, pattern)` that:
* Returns rows where the specified column contains the given pattern
* Handles missing values
* Prints the number of matching rows

**Sample Output:**

If your selected column contained matching text values, the following could be your output:
```
Found 1 records of color containing blue.
Found 1 records of product containing hat.
Found 1 records of color containing red.
```

````python
def filter_rows_matching_pattern(df, column, pattern):
    """Return rows whose column contains the pattern, ignoring missing values."""
    # Missing values are not matches, so they are left out of the result.
    matches = df[column].astype("string").str.contains(pattern, na=False, regex=False)
    result = df[matches]
    print(f"Found {len(result)} records of {column} containing {pattern}.")
    return result



````

````python
def create_df_categorical(df_source, column_name):
    """Return a DataFrame of category counts and the share of each category."""
    counts = df_source[column_name].value_counts()
    # Percentage is each count divided by the number of non-missing values.
    percentages = counts / counts.sum()
    return pd.DataFrame({"count": counts, "pct": percentages})



````

**Problem 17**

**Concept:** *Filtering records by category.*


**Task:**

Write a function `filter_weather(df, conditions, temp_min, temp_max)` that:
* Filters rows where:
  * conditions contain the given string
  * `tempmax` ≤ `temp_max`
  * `tempmin` ≥ `temp_min`
* Prints intermediate counts
* Returns the filtered DataFrame

**Sample Output:**

If your data matched the filtering conditions step by step, the following could be your output:
```
4 records with conditions Blue
3 records with conditions Blue and tempmax <= 50
2 records with conditions Blue and 20 <= temp <= 50
Found 2 records with conditions of Blue and temp between 20 and 50.
```

````python
def filter_weather(df, conditions, temp_min, temp_max):
    """Return rows matching a condition string and a temperature range."""
    # Keep rows whose conditions text contains the requested string.
    condition_match = df["conditions"].astype("string").str.contains(conditions, na=False, regex=False)
    df_conditions = df[condition_match]
    print(f"{len(df_conditions)} records with conditions {conditions}")
    # Also require the day's high to be at or below temp_max.
    df_max = df_conditions[df_conditions["tempmax"] <= temp_max]
    print(f"{len(df_max)} records with conditions {conditions} and tempmax <= {temp_max}")
    # Also require the day's low to be at or above temp_min.
    df_final = df_max[df_max["tempmin"] >= temp_min]
    print(f"{len(df_final)} records with conditions {conditions} and {temp_min} <= temp <= {temp_max}")
    print(
        f"Found {len(df_final)} records with conditions of {conditions} "
        f"and temp between {temp_min} and {temp_max}."
    )
    return df_final



````

**Concept:** *Creating DataFrame, standardizing values.*


**Task:**

Write a function `create_standardized_values(df_source, columns, row_index=None)` that:
* Creates a new DataFrame
* Standardizes values relative to a given row
* Uses the first row if none is provided
* Returns the new DataFrame

**Sample Output:**

If the following was your dataframe:
```
      high_value  low_value
2020        100        50
2021        120        55
2022        140        60
2023        160        70
2024        180        75
```
If you call the function
`create_standardized_values(df_example, ["high_value", "low_value"], 2020)`,
you should get the following as the output:
```
      high_value  low_value
2020         1.0       1.00
2021         1.2       1.10
2022         1.4       1.20
2023         1.6       1.40
2024         1.8       1.50

```

````python
def create_standardized_values(df_source, columns, row_index=None):
    """Return selected columns divided by the values in a reference row."""
    # Use the first row when no reference index is given.
    if row_index is None:
        row_index = df_source.index[0]
    reference = df_source.loc[row_index, columns]
    # Divide each value by the reference row so that row becomes 1.
    return df_source.loc[:, columns] / reference



````
