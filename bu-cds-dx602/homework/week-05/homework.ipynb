**Review Problem 1**

**Concept:** *Processing a list with a definite loop.*

**Task:**

Write a function `print_word_lengths(words)` that takes a list of strings and prints each word and its length on its own line in the format:

`word: length`

Use an element-based `for` loop. The function does not return anything.

**Sample Output:**
```
print_word_lengths(['apple', 'banana', 'kiwi'])
```
should print
```
apple: 5
banana: 6
kiwi: 4
````


`````python
def print_word_lengths(words):
    """Print each word in the list along with its length."""
    # Visit each word directly with an element-based loop.
    for word in words:
        # len() gives the number of characters in the word.
        print(f"{word}: {len(word)}")
`````

**Review Problem 2**

**Concept:** *Processing a list with a definite loop, using accumulator pattern.*

**Task:**

Write a function `total_length(words)` that returns the sum of the lengths of all words in a list using the accumulator pattern.

Write a second version using list comprehension.

**Sample Output:**

```
total_length(['data', 'science', 'rocks'])
```
should return `15`




`````python
def total_length(words):
    """Return the total number of characters in all words, using an accumulator."""
    total = 0
    # Add the length of each word to the running total.
    for word in words:
        total += len(word)
    return total


def total_length_2(words):
    """Return the total number of characters in all words, using a list comprehension."""
    # Build a list of word lengths, then add them up with sum().
    return sum([len(word) for word in words])
`````

**Review Problem 3**

**Concept:** *List processing, accumulation, decision statements, creating a new list as result.*

**Task:**

Write a function `union(a, b)` that returns a new list containing all unique elements from both lists.

Rules:

* Cannot use a set

* Cannot modify input lists

* Must create a new list

* Order does not matter

**Sample Output:**

```
union of [1] with [1,2] should return [1, 2]
union of [1, 3] with [2, 4, 5] should return [1, 2, 3, 4, 5]
union of [4, 5] [1, 2, 3, 4] should return [4, 5, 1, 2, 3]
```




`````python
def union(a, b):
    """Return a new list of the unique elements from lists a and b, without using a set."""
    result = []
    # Look at every element of a, followed by every element of b.
    for value in a + b:
        # Only add the value if it is not already in the result.
        if value not in result:
            result.append(value)
    return result
`````

**Problem 4**

**Concept:** *Using `set` to remove duplicates.*

**Task:**

Write a function `union_2(a, b)` that returns a list of unique elements by converting the combined list into a set and back into a list.

**Sample Output:**

```
union of [1] with [1,2] should return [1, 2]
union of [1, 3] with [2, 4, 5] should return [1, 2, 3, 4, 5]
union of [4, 5] [1, 2, 3, 4] should return [4, 5, 1, 2, 3]
```


`````python
def union_2(a, b):
    """Return a list of the unique elements from lists a and b, using a set."""
    # Combine both lists, convert to a set to drop duplicates, then convert back to a list.
    return list(set(a + b))
`````

**Problem 5**

**Concept:** *Process a string, find unique characters.*

**Task:**

Write a function `unique_letters(s)` that returns a list of unique characters in a string using:

* list comprehension

* set

**Sample Output:**
```
print(unique_letters('hello world'))
```
should return
```
unique letters: ['h', 'e', 'l', 'o', ' ', 'w', 'r', 'd']
```



`````python
def unique_letters(s):
    """Return the unique characters of s in first-seen order, using a list comprehension."""
    # Keep a character only if it does not appear earlier in the string (s[:i]).
    return [c for i, c in enumerate(s) if c not in s[:i]]


def unique_letters_2(s):
    """Return the unique characters of s, using a set (order is not preserved)."""
    # Convert the string to a set to drop duplicates, then back to a list.
    return list(set(s))
`````

Test case for problem 5

`````python
# test cases
print(f"unique letters: {unique_letters('hello world')}")
print(f"unique letters: {unique_letters_2('hello world')}")
`````

**Problem 7**

**Concept:** *Using dictionary data type, basic syntax.*

**Task:**

Write a function `word_score(word)` that returns the Scrabble score of a word using the dictionary `scrabble_scores`.

**Sample Output:**
```
print(word_score('cat'))
```
should return
```
cat: 5
```




`````python
def word_score(word):
    """Return the Scrabble score of word using the scrabble_scores dictionary."""
    score = 0
    # Add the point value of each letter to the running score.
    for letter in word:
        score += scrabble_scores[letter]
    return score
`````

**Problem 9**

**Concept:** *Using list data type, dictionary data type, loop, dictionary element access, accumulator pattern.*

**Task:**

Write a function `total_points(games_results, team_name)` that returns the total points scored by a team across all games.

**Sample Output:**
```
games_results = [
    {'A': 5, 'B': 7},
    {'A': 10, 'C': 3}
]
print(total_points(games_results, 'A'))
```
should return
```
total points for A: 15
```



`````python
def total_points(games_results, team_name):
    """Return the total points scored by team_name across all games."""
    total = 0
    # Visit each game dictionary in the list.
    for game in games_results:
        # Only count games the team actually played in.
        if team_name in game:
            total += game[team_name]
    return total
`````

**Concept:** *Using list data type, dictionary data type, sorting.*

**Task:**

Write a function `sort_games_by_score(games, team_name)` that returns a new list sorted by a team’s score in descending order.

**Sample Output:**

```
games = [
    {'A': 10, 'B': 5},
    {'A': 20, 'C': 3},
    {'A': 7, 'D': 14}
]
print_game_scores(sort_games_by_score(games, 'A'))
```
should return in descending order
```
A: 20
C: 3

A: 10
B: 5

A: 7
D: 14
```


`````python
def sort_games_by_score(games, team_name):
    """Return a new list of games sorted by team_name's score, highest first."""
    # sorted() returns a new list; the key looks up the team's score in each game
    # (0 if the team did not play), and reverse=True puts the highest score first.
    return sorted(games, key=lambda game: game.get(team_name, 0), reverse=True)
`````

**Problem 11**

**Concept:** *Using dictionary data type, loop, decision statements.*

**Task:**

Write a function `best_score(games, team_name)` that returns the game dictionary where the team scored the most points.

**Sample Output:**
```
games = [
    {'A': 8, 'B': 10},
    {'A': 15, 'C': 7},
    {'A': 12, 'D': 9}
]
print(best_score(games, 'A'))
```
should return
```
{'A': 15, 'C': 7}
```




`````python
def best_score(games, team_name):
    """Return the game dictionary in which team_name scored the most points."""
    best_game = None
    # Visit each game dictionary in the list.
    for game in games:
        # Skip games the team did not play in.
        if team_name in game:
            # Keep this game if it is the first one seen or beats the best so far.
            if best_game is None or game[team_name] > best_game[team_name]:
                best_game = game
    return best_game
`````

**Problem 12**

**Concept:** *Using list data type, dictionary data type, sorting.*

**Task:**

Write a function `best_word(words)` that returns the word with the highest Scrabble score.

**Sample Output:**
```
best_word(['dog', 'quiz', 'cat'])
```
should return `quiz`, because `quiz` has the highest scrabble score.




`````python
def best_word(words):
    """Return the word in words with the highest Scrabble score."""
    # Sort the words by their Scrabble score (word_score), highest first,
    # then take the first word.
    return sorted(words, key=word_score, reverse=True)[0]
`````
